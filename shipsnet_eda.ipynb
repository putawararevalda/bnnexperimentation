```python
# create a mapping of index (starting from 0) and file name
# use os

import os

def create_index_mapping(folder):
    index_mapping = {}
    for i, filename in enumerate(os.listdir(folder)):
        if filename.endswith('.png'):
            index_mapping[i] = filename
    return index_mapping
```

```python
ship_index_mapping = create_index_mapping(os.path.join(data_dir, 'ship'))
no_ship_index_mapping = create_index_mapping(os.path.join(data_dir, 'no_ship'))
```

```python
# load the images with pil
# show in matplotlib, the user inputs the folder name and the index of the image
from matplotlib import pyplot as plt
from PIL import Image
def show_image(folder_name, index):

    if folder_name not in ['ship', 'no_ship']:
        raise ValueError("folder_name must be 'ship' or 'no_ship'")
    elif folder_name == 'ship':
        index_mapping = ship_index_mapping[index]
    elif folder_name == 'no_ship':
        index_mapping = no_ship_index_mapping[index]

    img_path = f"{data_dir}/{folder_name}/{index_mapping}"
    img = Image.open(img_path)
    plt.imshow(img)
    plt.axis('off')
    plt.show()
```

```python
def show_images_in_grid(images):
    fig, axs = plt.subplots(1, len(images), figsize=(20, 5))
    for ax, (folder_name, index) in zip(axs, images):
        if folder_name not in ['ship', 'no_ship']:
            raise ValueError("folder_name must be 'ship' or 'no_ship'")
        elif folder_name == 'ship':
            index_mapping = ship_index_mapping[index]
        elif folder_name == 'no_ship':
            index_mapping = no_ship_index_mapping[index]

        img_path = f"{data_dir}/{folder_name}/{index_mapping}"
        img = Image.open(img_path)
        ax.imshow(img)
        ax.axis('off')
    plt.show()

show_images_in_grid([('ship', 1), ('no_ship', 5), ('no_ship', 1000), ('no_ship', 2002)])
```

```python
# i want to show 4 images in a grid of 1 row and 4 columns
# the file might be a combination of ship and no_ship
# the user inputs are like this: ['ship', 1), ('no_ship', 0), ('no_ship', 1000), ('no_ship', 2000)]
# give (1.a) (2.a) (2.b) (2.c) as the caption for each image

def show_images_with_captions(images_with_captions):
    fig, axs = plt.subplots(1, len(images_with_captions), figsize=(20, 5))
    for ax, (folder_name, index, caption) in zip(axs, images_with_captions):
        if folder_name not in ['ship', 'no_ship']:
            raise ValueError("folder_name must be 'ship' or 'no_ship'")
        elif folder_name == 'ship':
            index_mapping = ship_index_mapping[index]
        elif folder_name == 'no_ship':
            index_mapping = no_ship_index_mapping[index]

        img_path = f"{data_dir}/{folder_name}/{index_mapping}"
        img = Image.open(img_path)
        ax.imshow(img)
        ax.axis('off')
        ax.set_title(caption)
    plt.show()

show_images_with_captions([
    ('ship', 1, '1.a'),('no_ship', 0, '2.a'),
    ('no_ship', 1000, '2.b'), ('no_ship', 2000, '2.c')
])
```
